# Part 1 · The Map ⭐

<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 28px; border-radius: 18px; color: white; margin: 12px 0;">


### Three words people use as synonyms, which are not synonyms at all

**Artificial intelligence**, **machine learning**, **deep learning**. Treat them as the same thing and you will reach for a neural network when fifteen lines of `if` would do — or call a hand-tuned rule "AI" and wonder why feeding it more data changes nothing.

They are not rivals. They are **nested circles**, and what changes as you move inward is not the ambition. It is *how much of the thinking you hand over to the data.*


</div>

## 1.1 Three Nested Circles

<div style="background-color: #f3e5f5; border-left: 5px solid #9c27b0; padding: 15px; border-radius: 8px; margin: 10px 0;">

### 📖 The relationship

$$ \textbf{AI} \;\supset\; \textbf{ML} \;\supset\; \textbf{Deep Learning} $$

- **Artificial intelligence** is the broadest goal: *any* technique that makes a machine do something we would call intelligent — playing chess, planning a route, understanding speech. It includes approaches that never touch data at all.
- **Machine learning** is a *strategy* for reaching that goal: instead of writing the rules by hand, you write a program that **learns the rules from examples**.
- **Deep learning** is one *family* of models inside ML: neural networks with many layers stacked deep. "Deep" is literal — it means many layers. It is not a synonym for ML, and certainly not for AI.

</div>

### 🧠 Intuition — who is doing the thinking?

Picture the three circles as three answers to a single question: **when the program makes a decision, whose reasoning is it using?**

| Circle | Who supplies the rules | Who supplies the features |
|---|---|---|
| **AI** (outer) | you, by hand | you, by hand |
| **ML** (middle) | **the data** | you, by hand |
| **Deep learning** (inner) | **the data** | **the data** |

That table is the whole map. Moving inward hands over more of the work.

### 📊 What this shows


**The circles are nested, not side by side.** Every deep-learning system *is* a machine-learning system, and every machine-learning system *is* an AI system. The reverse is false — a hand-written rule engine is AI and nothing more.

**The one task in the middle** is deliberately the same in all three rings: *is this email spam?* Read outward-to-inward and watch what the human stops doing. In the outer ring you write the word list and the threshold. In the middle ring you still choose *what to measure* (word counts, link counts) but the data chooses the weights. In the inner ring you hand over the raw text and the network invents its own features.

**The ladder on the right** is the sentence to memorise: *in classical AI you write the rules; in ML the data picks the numbers; in deep learning the data picks the numbers **and** the features.*

**The common misread:** inner does not mean better. Most tabular problems are solved faster, cheaper and *more accurately* by a plain model in the middle ring. Moving inward costs data, compute, and interpretability — spend it only when the outer rings genuinely fail.


</div>

## 1.2 The Same Task in All Three Circles

Talk is cheap. Let us actually write the spam filter three times.

### Circle 1 — classical AI: you write the rule

Count how often the words *free, winner, prize, click* appear. Three or more, call it spam. That threshold of 3 is a number **a human picked**.


````python
# ============================================================
#  Circle 1 — a hand-written rule. Real AI, and it learns nothing.
# ============================================================
SPAM_WORDS = {"free", "winner", "prize", "click"}

def is_spam_by_rule(email_words):
    """Classical AI: the intelligence is yours, frozen into a constant."""
    hits = sum(1 for w in email_words if w in SPAM_WORDS)
    return hits >= 3                      # <- 3 chosen by a human, never revised

tests = [
    (["free", "prize", "click", "today"],       True),
    (["meeting", "notes", "attached"],          False),
    (["free", "lunch"],                         False),
    (["winner", "claim", "your", "prize"],      True),
]
print("Circle 1 — hand-written rule")
print("=" * 52)
for words, expected in tests:
    got = is_spam_by_rule(words)
    mark = "✅" if got == expected else "❌"
    print(f"  {mark} {str(words):<44} -> {'SPAM' if got else 'ham'}")

print()
print("Look at the last row. 'winner claim your prize' is obviously spam, but it")
print("contains only TWO listed words, so hits=2 < 3 and the rule waves it through.")
print("The rule is not broken -- it is doing exactly what a human told it. The human")
print("was wrong, and the rule has no way to find that out.")
print()
print("Notice what is missing: there is no .fit(). No data was consulted.")
print("Show this function a million emails and it makes the SAME mistakes")
print("on the millionth as on the first, because `3` never changes.")
````

### Circle 2 — machine learning: the data picks the numbers

Now we stop choosing the threshold. We hand the program **labelled examples** and let it find the rule itself.

Each email becomes two numbers — how many spam-ish words it contains ($x_1$) and how many links it has ($x_2$) — and a model looks for weights $w_1, w_2$ and a bias $b$ such that

$$ w_1 x_1 + w_2 x_2 + b $$

is comfortably positive for spam and comfortably negative for ham. **We never say what the weights should be.** The data settles them.


````python
# ============================================================
#  Circle 2 — the same task, learned from labelled examples
# ============================================================
from sklearn.linear_model import LogisticRegression

def make_emails(n, generator):
    """Synthetic inbox. Spam tends to carry LINKS more than spam-words --
    a fact the hand-written rule has no way to know."""
    is_spam = generator.random(n) < 0.4
    spam_words = np.where(is_spam, generator.poisson(2.0, n), generator.poisson(1.7, n))
    links      = np.where(is_spam, generator.poisson(3.6, n), generator.poisson(0.5, n))
    return np.c_[spam_words, links].astype(float), is_spam.astype(int)

# A held-out inbox neither approach ever trains on — the honest scoreboard.
X_test, y_test = make_emails(6000, np.random.default_rng(99))

# The hand-written rule, expressed on the same two features
def rule_predict(X):
    return ((X[:, 0] + X[:, 1]) >= 3).astype(int)   # the human's threshold

rule_accuracy = (rule_predict(X_test) == y_test).mean()

# The learned model, trained on just 300 labelled emails
X_train, y_train = make_emails(300, np.random.default_rng(1))
model = LogisticRegression().fit(X_train, y_train)
learned_accuracy = (model.predict(X_test) == y_test).mean()

w1, w2 = model.coef_[0]
b = model.intercept_[0]

print("Circle 2 — learned from data")
print("=" * 58)
print(f"  hand-written rule : {rule_accuracy:6.1%}")
print(f"  learned model     : {learned_accuracy:6.1%}   (trained on 300 emails)")
print()
print("The weights nobody typed in:")
print(f"  w1 (spam words) = {w1:+.3f}")
print(f"  w2 (links)      = {w2:+.3f}")
print(f"  b  (bias)       = {b:+.3f}")
print()
print(f"The data discovered that LINKS matter about {w2/w1:.0f}x more than spam-words.")
print("The hand-written rule weighted them equally. That is the whole gap.")
````

### ✅ Read those weights again

Nobody typed `+1.4` or `−3.2`. The model *derived* them by looking at labelled examples — and in doing so it discovered something the human rule-writer got wrong: in this inbox, **links are the real signal and spam-words are mostly noise**.

That is the shift that defines the middle circle. We stopped writing rules and started writing a **learner**. When the spammers change tactics, we do not edit code — we hand it new labelled emails and the weights move themselves.

</div>

### Circle 3 — deep learning: the data picks the features too

In Circle 2 we still had to *invent the columns* — someone decided to count spam-words and links. That is called **feature engineering**, and for raw pixels, audio, or unstructured text it is brutal, because nobody knows the right features to hand-craft.

Deep learning's move is to learn the features as well. A network stacks layers of simple units, each computing

$$ z = w \cdot x + b, \qquad a = \text{ReLU}(z) = \max(0, z) $$

so that early layers discover useful intermediate patterns and later layers combine them. "Deep" just means several such layers between input and output.

🔗 *We build exactly this machinery in Notebooks 2 and 4 — the $w \cdot x$ is a dot product (Notebook 2), and the learning of $w$ is the chain rule (Notebook 4).*


# Part 2 · The Fingerprint ⭐

<div style="background: linear-gradient(135deg, #11998e 0%, #38ef7d 100%); padding: 28px; border-radius: 18px; color: white; margin: 12px 0;">


### One test tells you whether you have crossed into machine learning

Not "does it feel smart." Not "does it use a neural network." The test is:

> **Does it get better when you feed it more labelled data?**

A hand-written rule stays flat forever, no matter how much data you show it. A learned model climbs. That difference is not a matter of degree — it is the boundary between the outer two circles, and you can *measure* it.


</div>

## 2.1 Measuring the Fingerprint

Here is the experiment. Take the same inbox. Train the learned model on 20 emails, then 50, then 100, then 300, up to 10,000 — and each time score it on the **same held-out test set**. Score the hand-written rule on that test set too.

The rule never sees the training data at all, which is precisely the point: it has nothing to learn *from*.


````python
# ============================================================
#  The fingerprint experiment: does more data help?
# ============================================================
train_sizes = [20, 50, 100, 300, 1000, 3000, 10000]
TRIALS = 12          # average over several random training sets, so the
                     # curve reflects the method and not one lucky draw

learned_curve, learned_spread = [], []
for n in train_sizes:
    scores = []
    for t in range(TRIALS):
        Xtr, ytr = make_emails(n, np.random.default_rng(1000 + t))
        if len(np.unique(ytr)) < 2:        # a tiny sample can be all one class
            continue
        clf = LogisticRegression().fit(Xtr, ytr)
        scores.append((clf.predict(X_test) == y_test).mean())
    learned_curve.append(np.mean(scores))
    learned_spread.append(np.std(scores))

rule_curve = [rule_accuracy] * len(train_sizes)   # flat, by construction

print(f"{'training emails':>16}{'hand rule':>12}{'learned':>10}")
print("-" * 40)
for n, r, l in zip(train_sizes, rule_curve, learned_curve):
    print(f"{n:>16,}{r:>11.1%}{l:>10.1%}")
print("-" * 40)
print(f"The rule never moves. The learner gains "
      f"{learned_curve[-1] - learned_curve[0]:+.1%} from 20 -> 10,000 emails,")
print(f"and beats the rule by {learned_curve[-1] - rule_accuracy:+.1%} at the end.")
````

````python
# ============================================================
#  Figure: the fingerprint — flat rule vs. climbing learner
# ============================================================
fig, ax = plt.subplots(figsize=(10.5, 5.4))

lo = np.array(learned_curve) - np.array(learned_spread)
hi = np.array(learned_curve) + np.array(learned_spread)
ax.fill_between(train_sizes, lo, hi, color=C_MODEL, alpha=0.15,
                label="learned model — spread over 12 runs")
ax.plot(train_sizes, learned_curve, "o-", color=C_MODEL, lw=2.6, ms=7,
        label="learned model (logistic regression)")
ax.plot(train_sizes, rule_curve, "s--", color=C_ERR, lw=2.6, ms=7,
        label="hand-written rule (no .fit)")

ax.set_xscale("log")
ax.set_xticks(train_sizes)
ax.set_xticklabels([f"{n:,}" for n in train_sizes])
ax.set_ylim(0.68, 0.95)
ax.yaxis.set_major_formatter(lambda v, p: f"{v:.0%}")

# annotate the gap at the right-hand edge
ax.annotate("", xy=(10000, learned_curve[-1]), xytext=(10000, rule_accuracy),
            arrowprops=dict(arrowstyle="<->", color=C_TRUE, lw=2))
ax.text(11500, (learned_curve[-1] + rule_accuracy) / 2,
        f"{learned_curve[-1] - rule_accuracy:+.1%}\nfrom learning",
        color=C_TRUE, fontsize=9.5, fontweight="bold", va="center")
ax.text(24, rule_accuracy + 0.006, "flat forever — no amount of data helps",
        color=C_ERR, fontsize=9, style="italic")

tidy(ax, "labelled emails used for training  (log scale)",
     "accuracy on the same held-out inbox",
     "The fingerprint of machine learning", legend=True)
plt.tight_layout(); plt.show()
````

<div style="background-color: #e7f3fe; border-left: 5px solid #2196F3; padding: 15px; border-radius: 8px; margin: 10px 0;">

### 📊 What this shows


**Axes.** Horizontal: how many labelled emails the model was allowed to learn from, on a log scale so 20 and 10,000 fit on one page. Vertical: accuracy measured on a **held-out** inbox of 6,000 emails that neither approach ever trained on.

**The red dashed line is flat, and it is flat by construction.** The hand-written rule has no `.fit()` step. There is no mechanism by which data could change it. Feeding it ten thousand emails and feeding it none produce byte-identical behaviour.

**The purple curve climbs, then levels off.** Both halves matter:

- *It climbs* — that rise is the fingerprint. More examples give the fitting procedure a better estimate of the weights, so the boundary lands closer to the truth.
- *It levels off* — the plateau is not a failure. It is the model reaching the ceiling of what **this model class** can express with **these two features**. More data cannot fix a model that is too simple; that needs a richer model (a step inward) or better features.

**The shaded band** is the spread across 12 independent training sets. Notice it is fat on the left and thin on the right: with 20 emails you are at the mercy of *which* 20 you got, and with 10,000 you are not. **Small training sets do not just give worse models — they give unpredictable ones.**

**The common misread:** the rule is not bad because it is simple. It is bad because a human guessed that spam-words and links should count equally, and in this inbox they do not. The learner beat it by discovering the right ratio, not by being more complicated.


</div>

<div style="background-color: #fff3cd; border-left: 5px solid #ffc107; padding: 15px; border-radius: 8px; margin: 10px 0;">

### ⚠️ The practical version of this test

Before you call anything "AI", ask: **is there a `.fit()`?**

If more labelled data cannot improve it, you have not crossed into machine learning — you have an automated rule, and that is a perfectly respectable thing to have. It is often the right thing to have. But it will not improve on its own, and no amount of data will make it.

</div>


# Part 3 · Three Shapes of Data ⭐

<div style="background: linear-gradient(135deg, #f093fb 0%, #f5576c 100%); padding: 28px; border-radius: 18px; color: white; margin: 12px 0;">


### Inside the ML circle, the *shape of your data* decides the paradigm

You do not choose supervised learning because it is popular. You choose it because your data came with answers attached. The data decides; you just have to recognise which case you are in.

## 3.1 The Three Paradigms

<div style="background-color: #e7f3fe; border-left: 5px solid #2196F3; padding: 15px; border-radius: 8px; margin: 10px 0;">

### 📐 Told apart by one question: what feedback do you get?

| Paradigm | What the data looks like | The feedback | Typical tasks |
|---|---|---|---|
| **Supervised** | every example carries the **right answer** | the exact target | classification (a category), regression (a number) |
| **Unsupervised** | examples with **no labels at all** | none — find structure yourself | clustering, dimensionality reduction |
| **Reinforcement** | no fixed answers, only a **reward** after acting | a score, possibly delayed | game playing, robotics, alignment |

</div>

Almost everything in this series — logistic regression, Naive Bayes, neural networks, the whole optimisation loop — lives in the **supervised** corner. It is worth knowing exactly which corner you are standing in, because the machinery differs.

### 📊 What this shows


**All three panels start from the same 90 points.** Only the *feedback* changes — which is exactly the claim: the paradigm is decided by the data you have, not by the algorithm you like.

**Left — supervised.** Each point arrives already coloured by its true class, so the model has a target to match and can measure exactly how wrong it is. The green dashed line is the boundary it learns. This is the setting for nearly everything in this series.

**Middle — unsupervised.** *The same points, with the colours taken away.* The algorithm must find structure with no notion of a right answer: here, two centres and the groups around them. Notice the honest ambiguity — nobody can tell it whether two clusters is the "correct" answer rather than three. There is no target to be wrong about.

**Right — reinforcement.** No labels exist at all. # Part 4 · A Model Is Just Numbers You Can Change ⭐⭐

<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 28px; border-radius: 18px; color: white; margin: 12px 0;">


### The loop that every model runs

In ordinary programming you write the rule yourself. To convert Celsius to Fahrenheit you type `F = C * 9/5 + 32`, because you knew that formula before you sat down.

Machine learning is for the enormous class of problems where **nobody knows the rule**. What formula turns the pixels of a photo into the word "cat"? What line separates a spam email from a real one? You cannot type these, because you do not know them.

So we stop trying to write the rule and **learn it from examples** — by guessing, measuring how wrong the guess is, nudging, and repeating. Four beats, forever.

## 4.1 The Simplest Possible Model

Let us make the arithmetic small enough to fit on a napkin. **Predict a student's exam score from the hours they studied.** We will guess the relationship is roughly a straight line:

$$ \hat{y} = w \cdot x + b $$

- $x$ — the input (hours studied)
- $\hat{y}$ — read "y-hat", the **prediction**
- $y$ — the **truth** we are trying to match
- $w$ — the **weight**, the line's slope
- $b$ — the **bias**, the line's intercept

<div style="background-color: #f3e5f5; border-left: 5px solid #9c27b0; padding: 15px; border-radius: 8px; margin: 10px 0;">

### 📖 What "a model" actually is

$w$ and $b$ are the model's **parameters** — the numbers the model is allowed to change. That is the whole definition.

Everything a model "learns" lives in numbers exactly like these. A network with 175 billion parameters is this same idea with billions of $w$'s stacked into matrices instead of one. When people say a model "knows" something, they mean *some of these numbers settled into a useful configuration.*

</div>

The model starts out knowing nothing, so we initialise the parameters to something arbitrary — say $w = 5$ and $b = 20$. A student studied 4 hours and actually scored 70. Let's see what our ignorant model says.


````python
# ============================================================
#  BEAT 2 — the model: two numbers and a multiplication
# ============================================================
def predict(hours, w, b):
    """The entire model. Two parameters, one line of arithmetic."""
    return w * hours + b          # ŷ = w·x + b

# The model starts out knowing nothing. These values are arbitrary.
w, b = 5.0, 20.0

x_one, y_one = 4.0, 70.0          # one student: studied 4 h, scored 70
y_hat = predict(x_one, w, b)

print("BEAT 2 — the model makes a guess")
print("=" * 46)
print(f"  parameters   w = {w},  b = {b}   (arbitrary — nobody chose these)")
print(f"  input        x = {x_one} hours studied")
print(f"  prediction   ŷ = {w} × {x_one} + {b} = {y_hat}")
print(f"  truth        y = {y_one}")
print()
print(f"  The model guessed {y_hat:.0f} when it should have said {y_one:.0f}.")
print(f"  It is off by {y_one - y_hat:.0f}.")
print()
print("That gap is the whole subject of the next beat.")
````

# Part 5 · Loss — Turning "Wrong" Into One Number ⭐⭐

<div style="background: linear-gradient(135deg, #fc466b 0%, #3f5efb 100%); padding: 28px; border-radius: 18px; color: white; margin: 12px 0;">


### To improve, the model needs a score it can push downward

"Off by 30" is a fact. A **loss** is that fact turned into a single number the machine can try to minimise — and, crucially, one whose *slope* it can follow.


</div>

## 5.1 Squared Error

The standard choice for numeric predictions is the **squared error**: take the gap, and square it.

$$ \text{loss} = (\hat{y} - y)^2 = (40 - 70)^2 = (-30)^2 = 900 $$

### 🧠 Why square it? Two reasons, both load-bearing

<div style="background-color: #e7f3fe; border-left: 5px solid #2196F3; padding: 15px; border-radius: 8px; margin: 10px 0;">

**1. Squaring kills the sign.** Being 30 too low and 30 too high are both errors of size 30. Without squaring they would be $-30$ and $+30$, and across a dataset they would **cancel** — a model that is wildly wrong in both directions would score a beautiful zero. Squaring makes every error count as a positive amount.

**2. Squaring makes a smooth bowl.** This is the reason that actually matters. A squared function is a parabola: it has a slope everywhere, that slope points toward the minimum, and it gets gentler as you approach the bottom. The next beat is entirely about *following a slope downhill*, and it needs a surface with slopes to follow. Absolute error $|\hat{y}-y|$ also kills the sign, but it has a sharp corner at zero where the slope is undefined.

</div>

Think of the loss as a **scoreboard**: it reads 900 today, and the entire job of training is to make that number smaller.


````python
# ============================================================
#  BEAT 3 — the loss: one number that says how wrong we are
# ============================================================
def squared_loss(y_hat, y):
    """The scoreboard. Lower is better; zero is perfect."""
    return (y_hat - y) ** 2

loss_now = squared_loss(y_hat, y_one)

print("BEAT 3 — measuring the wrongness")
print("=" * 46)
print(f"  prediction ŷ = {y_hat:.0f}")
print(f"  truth      y = {y_one:.0f}")
print(f"  gap          = {y_hat - y_one:+.0f}")
print(f"  loss         = ({y_hat - y_one:+.0f})² = {loss_now:.0f}")
print()

# Why squaring matters: watch the signs cancel without it
gaps = np.array([-30.0, +30.0, -10.0, +10.0])
print("  Four errors:", gaps.astype(int).tolist())
print(f"    plain sum of gaps   : {gaps.sum():>8.0f}   <- looks perfect. it is not.")
print(f"    sum of |gaps|       : {np.abs(gaps).sum():>8.0f}")
print(f"    sum of squared gaps : {(gaps ** 2).sum():>8.0f}   <- what we use")
print()
print("  The plain sum says this model is flawless. Squaring refuses to be fooled,")
print("  and unlike |gap| it is smooth everywhere — which the next beat needs.")
````

````python
# ============================================================
#  Figure: the miss, and the bowl that miss puts us on
# ============================================================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.9))

# ── left: the line, the truth, and the gap ──
hours = np.linspace(0, 8, 100)
ax1.plot(hours, predict(hours, w, b), color=C_MODEL, lw=2.6,
         label=f"model:  ŷ = {w:.0f}x + {b:.0f}")
ax1.scatter([x_one], [y_one], s=170, color=C_TRUE, zorder=5,
            edgecolor="white", lw=2, label=f"truth (4 h, {y_one:.0f})")
ax1.scatter([x_one], [y_hat], s=170, color=C_PRED, zorder=5,
            edgecolor="white", lw=2, label=f"prediction ({y_hat:.0f})")
ax1.annotate("", xy=(x_one, y_one), xytext=(x_one, y_hat),
             arrowprops=dict(arrowstyle="<->", color=C_ERR, lw=2.6))
ax1.text(x_one + 0.22, (y_one + y_hat) / 2, f"gap = {y_one - y_hat:.0f}",
         color=C_ERR, fontsize=11, fontweight="bold", va="center")
ax1.set_ylim(0, 90)
tidy(ax1, "hours studied (x)", "exam score", "The model misses by 30", legend=True)

# ── right: the loss as a function of w, with our position on it ──
w_grid = np.linspace(0, 25, 300)
loss_grid = (predict(x_one, w_grid, b) - y_one) ** 2
ax2.plot(w_grid, loss_grid, color=C_ERR, lw=2.6)
ax2.fill_between(w_grid, 0, loss_grid, color=C_ERR, alpha=0.08)
ax2.scatter([w], [loss_now], s=200, color=C_MODEL, zorder=5,
            edgecolor="white", lw=2)
ax2.annotate(f"we are here\nw = {w:.0f},  loss = {loss_now:.0f}",
             xy=(w, loss_now), xytext=(w + 2.2, loss_now + 700),
             fontsize=10, color=C_MODEL, fontweight="bold",
             arrowprops=dict(arrowstyle="->", color=C_MODEL, lw=1.8))
w_best = (y_one - b) / x_one
ax2.scatter([w_best], [0], s=220, marker="*", color=C_TRUE, zorder=5,
            edgecolor="white", lw=1.4)
ax2.text(w_best, 260, f"the bottom\nw = {w_best:.1f}", ha="center",
         fontsize=9.6, color=C_TRUE, fontweight="bold")
ax2.set_ylim(-200, 5200)
tidy(ax2, "the weight w", "loss  (ŷ − y)²",
     "Squaring turns the error into a bowl")

plt.tight_layout(); plt.show()
````

### 📊 What this shows


**Left — the miss, in the space the problem lives in.** Horizontal is hours studied, vertical is exam score. The purple line is our model with its arbitrary starting parameters. The green dot is what actually happened; the orange dot is what the model predicted. The red arrow between them is the error: **30 points of wrongness**.

**Right — the *same* situation, redrawn in the space training lives in.** This is the plot that matters, and the axes are different: horizontal is no longer *hours*, it is **the parameter $w$ itself**. Vertical is the loss you would get for each choice of $w$. Every point on this curve is a *different model*.

**Why is it a bowl?** Because the loss is a *squared* quantity, so as $w$ moves away from the value that fits, the penalty grows quadratically in both directions. The green star at the bottom marks the $w$ that would make the prediction exactly right — the loss there is 0.

**What to look for:** the purple dot is sitting high on the **left wall** of the bowl. That position carries two pieces of information that the next beat is entirely about: we are *far* from the bottom (the loss is large), and the ground under our feet slopes **down to the right** (so $w$ should increase). Training is nothing more than sliding that dot to the star.

**The common misread:** the right-hand bowl is not "the data". It is a picture of *how good every possible model would be*. The model never sees this curve — it only ever feels the slope directly under it, which is what makes the next beat necessary.


</div>

# Part 6 · The Update — Which Way Is Downhill? ⭐⭐⭐

<div style="background: linear-gradient(135deg, #11998e 0%, #38ef7d 100%); padding: 28px; border-radius: 18px; color: white; margin: 12px 0;">


### We know the loss is 900. How should we change w and b to make it smaller?

We could jiggle them at random and keep whatever helps. That works for two parameters and is hopeless for two hundred, never mind two billion.

So we ask a sharper question: **if we nudge w up a tiny bit, does the loss go up or down — and how fast?** That combination of *direction* and *rate* is the **gradient**.


</div>

## 6.1 The Gradient, and the Minus Sign

The gradient of the loss with respect to a parameter is the slope of the loss surface in that parameter's direction. For our squared error there is a tidy formula — and in Notebook 4 we will *derive* it rather than accept it:

$$ \frac{\partial L}{\partial w} = 2(\hat{y} - y)\cdot x, \qquad \frac{\partial L}{\partial b} = 2(\hat{y} - y) $$

The gradient points **uphill** — toward *increasing* loss. We want the opposite. So we step against it:

$$ \boxed{\; w \leftarrow w - \eta \cdot \frac{\partial L}{\partial w}, \qquad b \leftarrow b - \eta \cdot \frac{\partial L}{\partial b} \;} $$

That single line is **gradient descent**, and it is the engine under nearly all of modern machine learning. The $\eta$ (Greek *eta*) is the **learning rate**: how big a step to take.

<div style="background-color: #fff3cd; border-left: 5px solid #ffc107; padding: 15px; border-radius: 8px; margin: 10px 0;">

### ⚠️ The minus sign is not a detail

Drop it and you have *gradient ascent* — a model that reliably makes itself worse, as fast as possible. The gradient's job is to tell you which way is up; your job is to walk the other way.

</div>

Let us take exactly one step and watch the loss move.


````python
# ============================================================
#  BEAT 4 — the update: one step downhill
# ============================================================
eta = 0.001                     # the learning rate: how big a step

# the slopes of the loss with respect to each parameter
dL_dw = 2 * (y_hat - y_one) * x_one
dL_db = 2 * (y_hat - y_one)

print("BEAT 4 — taking one step")
print("=" * 54)
print(f"  ∂L/∂w = 2 × ({y_hat - y_one:+.0f}) × {x_one:.0f} = {dL_dw:+.0f}")
print(f"  ∂L/∂b = 2 × ({y_hat - y_one:+.0f})       = {dL_db:+.0f}")
print()
print("  Both are NEGATIVE, which reads: 'increasing w or b DECREASES the loss.'")
print("  Stepping against a negative gradient therefore means increasing them.")
print()

w_new = w - eta * dL_dw
b_new = b - eta * dL_db
y_hat_new = predict(x_one, w_new, b_new)
loss_new = squared_loss(y_hat_new, y_one)

print(f"  w: {w:.3f}  ->  {w_new:.3f}")
print(f"  b: {b:.3f}  ->  {b_new:.3f}")
print(f"  ŷ: {y_hat:.2f}  ->  {y_hat_new:.2f}      (target {y_one:.0f})")
print(f"  loss: {loss_now:.1f}  ->  {loss_new:.1f}")
print()
print(f"  The problem is not solved. It is nudged {loss_now - loss_new:.0f} in the right direction.")
print("  Do this ten thousand times and the line settles where the data wants it.")
````

## 6.2 The Learning Rate Decides Everything

The gradient tells you the **direction**. It says nothing about **how far the valley is** — and a step of size $\eta$ can sail straight past the bottom and land higher up the far wall.

This is the pitfall everyone hits first, so let us not describe it. Let us run it.


````python
# ============================================================
#  Three learning rates on the same bowl, same start
# ============================================================
def descend_w(eta, steps=12, w_start=5.0):
    """Gradient descent on w alone (b frozen), so we can draw it in 2-D."""
    ww, path = w_start, [(w_start, squared_loss(predict(x_one, w_start, b), y_one))]
    for _ in range(steps):
        g = 2 * (predict(x_one, ww, b) - y_one) * x_one
        ww = ww - eta * g
        path.append((ww, squared_loss(predict(x_one, ww, b), y_one)))
    return np.array(path)

runs = [("too small", 0.002, C_DATA),
        ("just right", 0.02, C_TRUE),
        ("too large", 0.075, C_ERR)]

fig, axes = plt.subplots(1, 3, figsize=(13.4, 4.4), sharey=True)
for ax, (name, e, col) in zip(axes, runs):
    path = descend_w(e)
    ax.plot(w_grid, loss_grid, color=C_GREY, lw=1.8, alpha=0.55)
    ax.plot(path[:, 0], path[:, 1], "o-", color=col, lw=1.9, ms=5.5, zorder=4)
    ax.scatter([path[0, 0]], [path[0, 1]], s=130, color=C_MODEL, zorder=6,
               edgecolor="white", lw=1.5)
    ax.scatter([w_best], [0], s=200, marker="*", color=C_TRUE, zorder=6,
               edgecolor="white", lw=1.2)
    ax.set_ylim(-300, 6000); ax.set_xlim(0, 25)
    ax.set_title(f"{name}\nη = {e}", fontsize=11)
    ax.set_xlabel("weight w")
    final = path[-1, 1]
    ax.text(0.5, 5400, f"loss after 12 steps: {final:,.0f}", fontsize=9.4,
            color=col, fontweight="bold")
    ax.spines[["top", "right"]].set_visible(False)
axes[0].set_ylabel("loss")
fig.suptitle("Same bowl, same start, same direction — only the step size differs",
             fontsize=12.5, fontweight="bold", y=1.02)
plt.tight_layout(); plt.show()

for name, e, _ in runs:
    p = descend_w(e)
    print(f"  η = {e:<6} {name:<11} loss 900 -> {p[-1,1]:>10,.1f}")
````

<div style="background-color: #e7f3fe; border-left: 5px solid #2196F3; padding: 15px; border-radius: 8px; margin: 10px 0;">

### 📊 What this shows


**Three panels, three learning rates, and everything else held identical** — same bowl, same starting point at $w=5$, same 12 steps, and in every panel the gradient points the *correct* way every single time.

**Left, $\eta$ too small.** The dots crawl. The direction was right at every step; there simply was not enough of it. After 12 steps the loss has barely moved. Nothing is broken — this model would get there eventually, wasting a great deal of compute on the way.

**Middle, $\eta$ just right.** The dots march down the wall and settle near the star. Notice they **bunch up** as they approach the bottom: nobody reduced the learning rate, but the *gradient itself* shrinks as the bowl flattens, so the steps shorten automatically. That self-braking is why a fixed $\eta$ can converge at all.

**Right, $\eta$ too large.** Watch the pattern carefully — the dots do not wander randomly, they **alternate**, jumping across the valley and landing higher on the opposite wall each time. Each step still points downhill from where it stands; it simply travels too far. That alternating signature in a loss curve is the fingerprint of a learning rate above the stable threshold.

**The lesson in one sentence:** gradient descent guarantees a *descent direction*, never a safe *distance* to travel along it. Choosing $\eta$ is a separate problem from computing the gradient — which is why every training recipe begins with a learning-rate sweep or a scheduler.


</div>

# Part 7 · The Whole Loop ⭐⭐⭐

<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 28px; border-radius: 18px; color: white; margin: 12px 0;">


### Four beats, eight lines, one working model

We have built data, model, loss and update separately. Now we assemble them — and run the loop on three real students to watch a loss of 856 fall to almost nothing.


</div>

## 7.1 The Skeleton, Assembled

Here is the whole thing. Read it as the four figure-boxes in order:

```python
w, b = 5, 20                    # 1. DATA is x, y; start with arbitrary parameters
for step in range(10000):
    y_hat = w * x + b           # 2. MODEL:  predict from data
    loss  = (y_hat - y)**2      # 3. LOSS:   measure the error
    dL_dw = 2*(y_hat - y)*x     #    the gradients
    dL_db = 2*(y_hat - y)
    w -= eta * dL_dw            # 4. UPDATE: nudge downhill
    b -= eta * dL_db            #    ... then REPEAT
```

<div style="background-color: #d4edda; border-left: 5px solid #28a745; padding: 15px; border-radius: 8px; margin: 10px 0;">

### ✅ This is not a toy version of training. It *is* training.

Swap the one-line model for a stack of matrix multiplications and nonlinear activations and you have a **neural network** — and the loop around it does not change a single character. Swap squared error for cross-entropy and you have a **classifier**; the loop is identical. Add momentum to the update and you have **Adam**; the loop is still identical.

Everything else in machine learning is a variation on *which* model, *which* loss, and *how* you take that step.

</div>

Now with three students instead of one, so the model has to find a compromise.


````python
# ============================================================
#  THE WHOLE LOOP — three students, 2000 steps
# ============================================================
# Three students: (hours studied, exam score)
X = np.array([2.0, 4.0, 6.0])
Y = np.array([45.0, 70.0, 88.0])

w, b = 5.0, 20.0          # back to knowing nothing
eta = 0.02
STEPS = 2000

history = []
for step in range(STEPS):
    y_hat = w * X + b                        # 2. MODEL
    errors = y_hat - Y
    loss = (errors ** 2).mean()              # 3. LOSS  (mean over the batch)
    history.append((w, b, loss))

    dL_dw = 2 * (errors * X).mean()          #    gradients, averaged over students
    dL_db = 2 * errors.mean()

    w -= eta * dL_dw                         # 4. UPDATE
    b -= eta * dL_db

history = np.array(history)

print("THE LOOP, RUNNING")
print("=" * 58)
print(f"{'step':>7}{'w':>10}{'b':>10}{'loss':>12}")
print("-" * 58)
for s in [0, 1, 2, 5, 20, 100, 500, STEPS - 1]:
    ws, bs, ls = history[s]
    print(f"{s:>7}{ws:>10.3f}{bs:>10.3f}{ls:>12.3f}")
print("-" * 58)
print(f"final model:  ŷ = {w:.3f}·x + {b:.3f}")
print(f"final loss :  {history[-1, 2]:.4f}   (started at {history[0, 2]:.1f})")
````

````python
# ============================================================
#  Figure: the line learning, and the loss falling
# ============================================================
fig = plt.figure(figsize=(13.2, 4.9))
gs = fig.add_gridspec(1, 2, width_ratios=[1.15, 1])

# ── left: a film-strip of the line at increasing steps ──
ax1 = fig.add_subplot(gs[0])
xs = np.linspace(0, 8, 50)
snapshots = [0, 2, 5, 15, 60, STEPS - 1]
shades = plt.cm.viridis(np.linspace(0.12, 0.92, len(snapshots)))
for (s, col) in zip(snapshots, shades):
    ws, bs, ls = history[s]
    ax1.plot(xs, ws * xs + bs, color=col, lw=2.0,
             label=f"step {s:>4}   loss {ls:7.1f}")
ax1.scatter(X, Y, s=170, color=C_TRUE, zorder=6, edgecolor="white", lw=2,
            label="the three students")
ax1.set_ylim(0, 100)
tidy(ax1, "hours studied", "exam score", "The line finds the data")
ax1.legend(fontsize=8.2, loc="lower right")

# ── right: the loss curve ──
ax2 = fig.add_subplot(gs[1])
ax2.plot(history[:, 2], color=C_ERR, lw=2.4)
ax2.set_yscale("log")
ax2.axhline(history[-1, 2], color=C_TRUE, ls="--", lw=1.6)
ax2.text(STEPS * 0.42, history[-1, 2] * 1.7,
         f"floor ≈ {history[-1, 2]:.2f} — the best a straight line can do",
         fontsize=9.2, color=C_TRUE, fontweight="bold")
for s, col in zip([0, 2, 5, 15, 60], shades):
    ax2.scatter([s], [history[s, 2]], s=55, color=col, zorder=5,
                edgecolor="white", lw=1.1)
tidy(ax2, "training step", "loss  (log scale)", "The scoreboard falls")

plt.tight_layout(); plt.show()
````

<div style="background-color: #e7f3fe; border-left: 5px solid #2196F3; padding: 15px; border-radius: 8px; margin: 10px 0;">

### 📊 What this shows


**Left — the same model at six moments in its training**, from ignorant (darkest) to trained (brightest). At step 0 the line is our arbitrary guess, passing nowhere near the three green students. By step 15 it has the slope roughly right. By step 2000 it is the best straight line that exists for this data.

Look at the **spacing between the lines**: enormous between steps 0 and 2, barely visible between 60 and 2000. Nothing changed the learning rate — the *gradient* shrank as the fit improved, so the steps shortened by themselves.

**Right — the same run as a scoreboard**, loss against step on a log axis. The near-straight initial descent on a log scale means the loss is falling by a constant *factor* per step, which is the signature of gradient descent on a well-behaved bowl.

**The most important feature of this figure is the dashed green floor.** The loss does not reach zero, and it never will. Three points do not lie on any straight line, so *no* setting of $w$ and $b$ can fit them all. What gradient descent found is the **compromise** that makes the total squared error as small as possible — and that residual is not a bug, it is the honest statement that a line is the wrong shape for perfectly fitting three arbitrary points.

**The common misread:** a loss that stops falling does not mean training failed or the learning rate is wrong. Here it means the model has reached the best it can be. Distinguishing "converged" from "stuck" is a real skill, and the first check is always: *could this model class even represent the answer?*


</div>

# Part 8 · Generalisation — Why a Rule Beats a Lookup Table ⭐⭐⭐

<div style="background: linear-gradient(135deg, #11998e 0%, #38ef7d 100%); padding: 28px; border-radius: 18px; color: white; margin: 12px 0;">


### The model did not memorise. It could not have.

It is tempting to think the loop above just stored `(4 hours → 70)` in a table. It did not, and here is the proof: **a line has only two numbers to remember.** Two numbers cannot hold three students.

What gradient descent found instead was a *compromise* — and that compromise can answer a question about a student it has never seen. That ability is called **generalisation**, and it is the entire point.


</div>

````python
# ============================================================
#  The model answers about a student it never saw
# ============================================================
print("What the model was shown")
print("=" * 52)
for xi, yi in zip(X, Y):
    print(f"   {xi:.0f} hours  ->  scored {yi:.0f}      predicted {predict(xi, w, b):6.2f}"
          f"   (off by {predict(xi, w, b) - yi:+5.2f})")

print()
print("It misses EVERY training point a little. No line can pass through all three.")
print(f"Total mean squared error: {((predict(X, w, b) - Y) ** 2).mean():.3f}  — not zero, and cannot be.")
print()
print("But now ask it about a student it has never seen:")
print("=" * 52)
for hours in [3.0, 5.0, 7.0]:
    print(f"   {hours:.0f} hours  ->  predicts {predict(hours, w, b):.1f}")
print()
print("Nobody stored '5 hours'. The model has exactly two numbers in it:")
print(f"   w = {w:.3f}   b = {b:.3f}")
print("Those two numbers ARE the learned rule. Predicting well on unseen")
print("inputs is called GENERALISATION, and it is the goal — not perfect recall.")
````

````python
# ============================================================
#  Figure: a compromise line that answers about the unseen
# ============================================================
fig, ax = plt.subplots(figsize=(10.4, 5.4))
xs = np.linspace(0.5, 8.5, 100)

ax.plot(xs, predict(xs, w, b), color=C_MODEL, lw=2.8,
        label=f"the learned rule:  ŷ = {w:.2f}x + {b:.2f}")

# the residual to every training point — the model is wrong about all of them
for xi, yi in zip(X, Y):
    yh = predict(xi, w, b)
    ax.plot([xi, xi], [yi, yh], color=C_ERR, lw=2.2, zorder=3)
    ax.text(xi + 0.12, (yi + yh) / 2, f"{yh - yi:+.1f}", color=C_ERR,
            fontsize=8.8, va="center", fontweight="bold")
ax.scatter(X, Y, s=180, color=C_TRUE, zorder=6, edgecolor="white", lw=2,
           label="the three students it saw")

# the unseen student
x_new = 5.0
y_new = predict(x_new, w, b)
ax.scatter([x_new], [y_new], s=230, marker="*", color=C_PRED, zorder=7,
           edgecolor="white", lw=1.8, label="a student it never saw (5 h)")
ax.annotate(f"never seen before\npredicted {y_new:.1f}",
            xy=(x_new, y_new), xytext=(x_new - 2.5, y_new + 17),
            fontsize=10, color=C_PRED, fontweight="bold",
            arrowprops=dict(arrowstyle="->", color=C_PRED, lw=1.9))

ax.set_ylim(30, 105)
tidy(ax, "hours studied", "exam score",
     "Wrong about every point it saw — and useful about one it did not", legend=True)
plt.tight_layout(); plt.show()
````

<div style="background-color: #e7f3fe; border-left: 5px solid #2196F3; padding: 15px; border-radius: 8px; margin: 10px 0;">

### 📊 What this shows


**The red bars are the model's mistakes on its own training data**, and there are three of them because the model is wrong about all three students it was shown. A memoriser would have zero red bars. This model cannot memorise: it has exactly two numbers in it, and two numbers cannot store three arbitrary points.

**The orange star is the payoff.** No student studying 5 hours appears anywhere in the training set, yet the model answers confidently and sensibly — because what it learned is not a table of answers but a **rule**, and rules extend to inputs they have not met.

**Why the line sits exactly where it does.** Gradient descent settled it at the position that makes the *total* squared error across all three students as small as possible. Push the line up to touch the top point and the other two get worse by more than you gained. That balance point is the compromise, and it is what "fitting" means.

**What to look for:** the three red bars point in *different directions* — the model is too high on some students and too low on others. A model that was consistently wrong in one direction would still have room to improve by shifting; this one has genuinely run out of improvements available to a straight line.

**The common misread:** those residuals are not failure. Demanding they go to zero is exactly how you get the disaster in Part 9.


</div>

# Part 9 · When It Breaks — Overfitting ⭐⭐⭐⭐

<div style="background: linear-gradient(135deg, #fc466b 0%, #3f5efb 100%); padding: 28px; border-radius: 18px; color: white; margin: 12px 0;">


### Give a model enough parameters and it *will* memorise

The residuals in Part 8 existed because a line has only two numbers. Give the model nine numbers and it can pass through all nine training points **exactly** — driving the training loss to zero.

And that is when it becomes useless. Watch.


</div>

````python
# ============================================================
#  Overfitting: the training loss lies to you
# ============================================================
# A more realistic dataset: 9 students, a gently curved truth, and noise
# (real exam scores are never exactly on a curve).
def true_relationship(hours):
    return 45 + 9.5 * hours - 0.42 * hours ** 2

gen = np.random.default_rng(SEED)
x_train = np.sort(gen.uniform(1, 9, 9))
y_train = true_relationship(x_train) + gen.normal(0, 3.0, 9)

# a large "future intake" the model never trains on — the honest scoreboard
x_test = np.linspace(1, 9, 200)
y_test = true_relationship(x_test) + gen.normal(0, 3.0, 200)

degrees = [1, 2, 3, 5, 8]
rows = []
for d in degrees:
    coeffs = np.polyfit(x_train, y_train, d)     # fit a degree-d polynomial
    fitted = np.poly1d(coeffs)
    mse_train = ((fitted(x_train) - y_train) ** 2).mean()
    mse_test  = ((fitted(x_test)  - y_test)  ** 2).mean()
    rows.append((d, d + 1, mse_train, mse_test, fitted))

print("Fitting the SAME 9 students with models of growing flexibility")
print("=" * 68)
print(f"{'degree':>7}{'parameters':>12}{'train MSE':>13}{'test MSE':>15}")
print("-" * 68)
for d, npar, tr, te, _ in rows:
    flag = "  <- memorised!" if tr < 0.01 else ""
    print(f"{d:>7}{npar:>12}{tr:>13.2f}{te:>15,.1f}{flag}")
print("-" * 68)
print("Read the two number columns against each other. Training error falls")
print("all the way to ZERO. Test error falls, bottoms out, then EXPLODES.")
````

````python
# ============================================================
#  Figure: under-fit, just right, and memorised
# ============================================================
fig = plt.figure(figsize=(13.4, 7.6))
gs = fig.add_gridspec(2, 3, height_ratios=[1, 0.95], hspace=0.42, wspace=0.25)

showcase = [(1, "TOO SIMPLE\na straight line", C_DATA),
            (2, "JUST RIGHT\nmatches the truth", C_TRUE),
            (8, "MEMORISED\nzero training error", C_ERR)]
xs = np.linspace(1, 9, 400)

for col, (d, title, colr) in enumerate(showcase):
    ax = fig.add_subplot(gs[0, col])
    fitted = dict((r[0], r[4]) for r in rows)[d]
    ax.plot(xs, true_relationship(xs), color=C_GREY, lw=2, ls="--",
            label="the true relationship")
    ax.plot(xs, fitted(xs), color=colr, lw=2.6, label=f"degree {d} fit")
    ax.scatter(x_train, y_train, s=95, color=C_TRUE, zorder=6,
               edgecolor="white", lw=1.6, label="9 students seen")
    ax.set_ylim(20, 105)
    ax.set_title(title, fontsize=11, color=colr)
    ax.set_xlabel("hours studied")
    if col == 0:
        ax.set_ylabel("exam score")
    ax.spines[["top", "right"]].set_visible(False)
    if col == 2:
        ax.legend(fontsize=7.6, loc="lower center")

# ── bottom: the two error curves that tell the real story ──
ax = fig.add_subplot(gs[1, :])
ds     = [r[0] for r in rows]
tr_mse = [r[2] for r in rows]
te_mse = [r[3] for r in rows]
# a training error of exactly 0 is -inf on a log axis: pin it to the floor
FLOOR = 2e-3
tr_plot = [max(v, FLOOR) for v in tr_mse]
ax.plot(ds, tr_plot, "o-", color=C_MODEL, lw=2.6, ms=8, label="training error — what you can see")
if tr_mse[-1] < FLOOR:
    ax.scatter([ds[-1]], [FLOOR], s=190, marker="v", color=C_MODEL, zorder=7,
               edgecolor="white", lw=1.6)
    ax.annotate("training error = EXACTLY 0  (below the log axis)",
                xy=(ds[-1], FLOOR), xytext=(ds[-1] - 3.4, FLOOR * 6),
                fontsize=9.6, color=C_MODEL, fontweight="bold",
                arrowprops=dict(arrowstyle="->", color=C_MODEL, lw=1.8))
ax.plot(ds, te_mse, "s-", color=C_ERR,   lw=2.6, ms=8, label="test error — what actually matters")
ax.set_yscale("log")
best = ds[int(np.argmin(te_mse))]
ax.axvline(best, color=C_TRUE, ls="--", lw=1.8)
ax.text(best + 0.12, 8e3, f"sweet spot\ndegree {best}", color=C_TRUE,
        fontsize=10, fontweight="bold")
ax.fill_between([best, 8.4], 1e-3, 1e7, color=C_ERR, alpha=0.06)
ax.text(6.6, 4e4, "OVERFITTING\ntraining error keeps falling,\nreal performance collapses",
        color=C_ERR, fontsize=10, fontweight="bold", ha="center")
ax.set_xticks(ds); ax.set_xlim(0.7, 8.4); ax.set_ylim(1e-3, 1e7)
tidy(ax, "model flexibility (polynomial degree)", "mean squared error (log)",
     "The training score is not the score that matters", legend=True)

plt.tight_layout(); plt.show()
````

### 📊 What this shows


**Top row — the same nine students, fitted three ways.** The grey dashed curve is the *true* relationship the data came from; the coloured curve is what each model learned.

- **Degree 1** cannot bend at all. It misses the curvature systematically — too low at the ends, too high in the middle. This is **under-fitting**: the model class is not rich enough to express the truth.
- **Degree 2** lands almost exactly on the grey dashed line. It recovered the real shape from noisy samples, which is what we wanted.
- **Degree 8** passes through **every single training point exactly**, and to do so it thrashes wildly between them. Between the points it is not merely inaccurate, it is absurd.

**Bottom — the diagnosis, and the most important plot in this notebook.** Two error curves against model flexibility, log scale.

The **purple** curve is the training error — the number your code prints during training. It falls monotonically and reaches **exactly zero** at degree 8. On this evidence alone, degree 8 is the best model you have ever built.

The **red** curve is the error on students the model never saw. It falls, bottoms out at the green line, and then **explodes by seven orders of magnitude**. Degree 8 is in fact the worst model you have ever built.

**The gap between those two curves is the whole of applied machine learning.** A model with enough parameters can always memorise; memorising is not learning, and the training loss cannot tell you the difference. Only held-out data can.

**The common misread:** overfitting is not caused by training too long or by a bad learning rate. Here every model was fitted exactly, in closed form, with no optimiser at all. It is caused by **too much flexibility for the amount of data you have** — and the cure is a simpler model, more data, or a penalty on complexity (regularisation).

### One equation has survived every era of this field

Strip away seventy years of jargon — perceptrons, expert systems, kernels, deep nets, transformers — and the learning rule at the bottom has barely moved:

$$ \boldsymbol{\theta} \;\leftarrow\; \boldsymbol{\theta} \;-\; \eta \, \nabla L(\boldsymbol{\theta}) $$

Parameters, a loss, its gradient, a learning rate, and a subtraction. **You have already written this.** It is the update beat from Part 6, in general notation.


</div>

## 10.1 If the Loop Is So Old, Why Did Everything Happen Recently?

Bayes died in 1761. Gradient descent is nineteenth-century calculus. Backpropagation was complete by the late 1980s. Yet the systems that write code and draft email arrived in the 2010s.

**The maths did not change. Three other things did.**

<div style="background-color: #e7f3fe; border-left: 5px solid #2196F3; padding: 15px; border-radius: 8px; margin: 10px 0;">

| What changed | The story |
|---|---|
| **Depth** | A single neuron draws one straight line, so it famously cannot solve XOR — a 1969 result that helped trigger the first AI winter. The fix was never a new idea; it was **stacking a hidden layer**, and a way to assign blame to it (backpropagation, Notebook 4). |
| **Data** | ImageNet was ~1.2 million labelled images across 1,000 categories — a training set that simply did not exist in the winter years. The quality of $\nabla L$ depends entirely on how many examples you average it over. |
| **Compute** | A network's forward and backward passes are, underneath, **dense matrix multiplications** (Notebook 2). A GPU does thousands of those independent dot products in parallel. In 2012 AlexNet trained on consumer GPUs and won ImageNet by a margin that silenced the room. |

</div>

The winters were not periods of wrong ideas. They were periods of **starvation** — the recipe was complete and the fuel was missing.

There has been arguably **one genuinely new idea** in this toolbox in seventy years: the parallel, all-pairs mixing of **attention** (2017), which happens to map perfectly onto the parallel hardware that had just arrived. Everything else that looked miraculous was old mathematics finally given enough data and enough compute to work.

The optimisation loop is a constant; **the architecture fed into it still matters enormously.**

# 🏆 Cheat Sheet

## The four-beat loop

```
DATA  ──►  MODEL  ──►  LOSS  ──►  UPDATE  ──┐
   examples   ŷ = w·x+b   (ŷ−y)²    θ ← θ − η∇L │
      ▲                                        │
      └──────────────── repeat ◄───────────────┘
```

## The vocabulary

| Term | Meaning |
|---|---|
| $x$ | input / feature |
| $y$ | the true answer (the **target**) |
| $\hat{y}$ | the model's **prediction** |
| $w$, $b$ | **parameters** — the numbers the model may change |
| $\theta$ | all the parameters, bundled |
| $L$ | the **loss** — one number saying how wrong |
| $\nabla L$ | the **gradient** — which way is uphill, and how steep |
| $\eta$ | the **learning rate** — how big a step |
| **generalisation** | performing well on inputs never seen in training |
| **overfitting** | fitting the training data so closely that real performance collapses |

## The three circles

| Circle | Rules from | Features from |
|---|---|---|
| classical AI | you | you |
| machine learning | **data** | you |
| deep learning | **data** | **data** |

## The three paradigms

| | Data has | You learn |
|---|---|---|
| **supervised** | inputs **and** answers | a mapping input → answer |
| **unsupervised** | inputs only | structure |
| **reinforcement** | a reward after acting | a policy |

# ⚠️ Common Pitfalls to Avoid

1. **Calling a frozen rule "AI that will improve."** If there is no `.fit()`, more data changes nothing. Run the Part 2 measurement.
2. **Judging a model on its training data.** Part 9 showed a model with *zero* training error and catastrophic real performance. Hold data back before you start.
3. **Reading a falling loss curve as success.** It proves the *optimiser* works. It says nothing about whether the objective was right or whether the model generalises.
4. **Assuming a plateau means failure.** In Part 7 the loss stopped falling because a straight line had reached its best. "Converged" and "stuck" look identical on the plot; ask whether the model class could even represent the answer.
5. **Forgetting the minus sign.** $\theta \leftarrow \theta + \eta\nabla L$ climbs the loss as fast as possible.
6. **Blaming the direction when the step size is at fault.** In Part 6 the diverging run pointed downhill every single time. The direction was right; $\eta$ was too big.
7. **Reaching inward too fast.** Most tabular problems are solved better, faster and cheaper by a plain model in the middle circle than by a neural network.
8. **Treating unsupervised as "supervised without the labelling work."** They answer different questions. There is no "correct" clustering to be wrong about.
9. **Averaging away the variance.** With small training sets you get worse *and* less predictable models — report the spread, not just the mean.
